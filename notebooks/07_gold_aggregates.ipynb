```python
# ==========================================
# 07 - GOLD BUSINESS AGGREGATES
# ==========================================

# ------------------------------------------
# 1. DAILY LOAN SUMMARY
# ------------------------------------------

spark.sql("""
CREATE OR REPLACE TABLE
mortgage_catalog.gold.agg_daily_loans AS

SELECT
    application_date,
    loan_status,
    COUNT(*) AS total_loans,
    SUM(loan_amount) AS total_loan_amount,
    AVG(loan_amount) AS avg_loan_amount

FROM mortgage_catalog.gold.fact_loan

GROUP BY
    application_date,
    loan_status
""")

print("agg_daily_loans → SUCCESS")


# ------------------------------------------
# 2. MONTHLY PAYMENT COLLECTION
# ------------------------------------------

spark.sql("""
CREATE OR REPLACE TABLE
mortgage_catalog.gold.agg_monthly_collection AS

SELECT
    YEAR(payment_date) AS payment_year,
    MONTH(payment_date) AS payment_month,
    payment_status,
    COUNT(*) AS total_payments,
    SUM(payment_amount) AS total_collection

FROM mortgage_catalog.gold.fact_payment

GROUP BY
    YEAR(payment_date),
    MONTH(payment_date),
    payment_status
""")

print("agg_monthly_collection → SUCCESS")


# ------------------------------------------
# 3. CUSTOMER LOAN SUMMARY
# ------------------------------------------

spark.sql("""
CREATE OR REPLACE TABLE
mortgage_catalog.gold.agg_customer_loan_summary AS

SELECT
    C.customer_id,
    C.customer_name,
    C.city,

    COUNT(L.loan_id) AS total_loans,

    COALESCE(
        SUM(L.loan_amount),
        0
    ) AS total_loan_amount

FROM mortgage_catalog.gold.dim_customer C

LEFT JOIN mortgage_catalog.gold.fact_loan L
    ON C.customer_id = L.customer_id

GROUP BY
    C.customer_id,
    C.customer_name,
    C.city
""")

print("agg_customer_loan_summary → SUCCESS")


print("\\nALL GOLD AGGREGATES COMPLETED ✅")
```
